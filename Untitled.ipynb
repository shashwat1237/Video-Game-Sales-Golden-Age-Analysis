```python
def convert_z_score(a):
    m=a.mean()
    s=a.std()
    return (a-m)/s
```

```python
def remove_outliers(a):
    q25=a.quantile(0.25)
    q75=a.quantile(0.75)
    IQR=q75-q25
    lower_limit=q25-(1.5*IQR)
    upper_limit=q75+(1.5*IQR)
    return a[(a>=lower_limit)&(a<=upper_limit)]
```

```python
top_platforms=df.groupby("Platform")["Global_Sales"].mean().sort_values(ascending=False).head(10).index
top_publishers=(df.groupby("Publisher")["Global_Sales"].mean().sort_values(ascending=False)).head(10).index
```

```python
# Create two new features:
# Platform_Group = 1 if the game's platform is in the top 10 platforms, else 0
# Publisher_Group = 1 if the game's publisher is in the top 10 publishers, else 0
```

```python
df["Platform_Group"]=df["Platform"].isin(top_platforms).astype(int)
df["Publisher_Group"]=df["Publisher"].isin(top_publishers).astype(int)
```

```python
def cut_into_bins(a,no):
    minn=a.min()
    maxx=a.max()
    step=(maxx-minn)/no
    bin_edges=[minn+(step*i) for i in range(no)]+[maxx]
    return pd.cut(a,bins=bin_edges,labels=list(range(no)),include_lowest=True)
```

```python
#The above function divides series into bins on the basis of equal numerical intervals but the function below (cut_into_bins2) divides 
#on the basis of quantiles
```

```python
def cut_into_bins2(a,no):
    return pd.qcut(a,q=no,labels=[0,1,2])
df["Year_bins"]=cut_into_bins2(df["Year"],3)
df["Year_bins"].value_counts()
```

```python
# Define the categorical features and their index positions
# Build a ColumnTransformer that applies Ordinal Encoding to these categorical columns
# Create a Pipeline that first preprocesses the data, then trains a Random Forest Regressor
# Prepare feature matrix X and target Y (log-transformed sales)
# Split the dataset into training and testing sets for model evaluation
```

```python
categorical_features=["Platform_Group","Genre","Publisher_Group","Year_bins"]
categorical_indices=list(range(len(categorical_features)))
ct=ColumnTransformer(transformers=[
    ("cat",OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1),categorical_features)
],remainder="drop")
pipe=Pipeline(steps=[
    ("preprocessor",ct),
    ("model",RandomForestRegressor(n_estimators=100,
                                  max_depth=20,
                                  random_state=42))
])
X=df[["Platform_Group","Genre","Publisher_Group","Year_bins"]]
Y=df["Global_sales_log"]
#Y=cut_into_bins(Y)
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=42)
```

```python

pipe.fit(X_train,Y_train)
Y_pred=pipe.predict(X_test)
r2=r2_score(Y_test,Y_pred)
print(r2)
```

```python
# Define a search space of Random Forest hyperparameters (number of trees + max depth)
# Set up a RandomizedSearchCV to try 25 random combinations using 10-fold cross-validation
# Optimize for the best R2 score
```

```python
param_grid={"model__n_estimators":[5,10,25,50,75,100,125,150,175,200,300,400],
           "model__max_depth":[2,5,6,7,8,9,10,12,15,17,20,22,25,27,30]}
grid_search=RandomizedSearchCV(estimator=pipe,
                        cv=10,
                        n_iter=25,
                        scoring="r2",
                        param_distributions=param_grid)
```

```python
grid_search.fit(X_train,Y_train)
```

```python
grid_search.best_score_
```

```python
# Re-split the data, this time stratifying Y for classification balance
# Build a second pipeline (pipe1) using RandomForestClassifier instead of Regressor
# This is for turning the sales prediction into a classification task
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=42,stratify=Y)

pipe1=Pipeline(steps=[
    ("preprocessor",ct),
    ("model",RandomForestClassifier(n_estimators=100,
                                  max_depth=20,
                                  random_state=42))
])
```

```python
# Use the same hyperparameter search space for the classifier model
# Perform RandomizedSearchCV to find the best estimator for accuracy
# Runs 25 random combinations across 10-fold CV
```

```python
param_grid={"model__n_estimators":[5,10,25,50,75,100,125,150,175,200,300,400],
           "model__max_depth":[2,5,6,7,8,9,10,12,15,17,20,22,25,27,30]}
grid_search1=RandomizedSearchCV(estimator=pipe1,
                        cv=10,
                        n_iter=25,
                        scoring="accuracy",
                        param_distributions=param_grid)
```

```python
grid_search1.fit(X_train,Y_train)
```

```python
grid_search1.best_score_
```

```python
Y_pred=grid_search1.predict(X_test)
print(accuracy_score(Y_test,Y_pred))
```

```python
best_estimator=grid_search1.best_estimator_
```
